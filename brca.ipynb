### Exploration over [dataset](./data/dataset.csv) of breast cancer gene (BRCG)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
pd.set_option('display.max_columns', 1000)
pd.set_option('display.max_rows', 1000)
```

```python
df = pd.read_csv("./data/dataset.csv", sep='\t')

df.columns = df.columns.str.replace(r'\xa0', ' ', regex=True).str.strip()

pd.concat([df.head(10), df.tail(10)])
```

```python
df.describe()
```

```python
# dbNSFP pathogenicity predictors
DBNSFP_PREDICTOR_COLS = [c for c in df.columns if c.startswith("dbNSFP_") and c.endswith("_pred")]

print(len(DBNSFP_PREDICTOR_COLS))

# '.' encodes missing values 
# so we need to pass then as nan cause pandas' dumb :D
missing_predicators = (
    df[DBNSFP_PREDICTOR_COLS]
    .replace('.', np.nan)
    .isna()
    .mean()
    .sort_values(ascending=False))

print(missing_predicators[missing_predicators > 0].to_string())

```

```python
predicators_vocab = { col: sorted(df[col].astype(str).unique()) for col in DBNSFP_PREDICTOR_COLS }

print('\n\n')
for col, vocab in predicators_vocab.items():
    print(f"{col:<32} {vocab}")
```

O **dbNSFP** (database for nonsynonymous SNP's functional predicators) é um banco de dados
que agrega, pra cada variante _missense_ conhecida, as predições de **patogenicidade** produzida por ferramentas computacionais independentes.
Como cada ferramenta é treinada/calibrada (etc etc) com ferramentas distintas com trade-offs diferentes, isoladamente não são perfeitas, mas em conjunto, quando a maioria converge pra um _deletério_, a evidência é mais sólida do que qualquer preditor isolado

| Code | Meaning |
| ---- | ------- |
| D | Damaging |
| T | Tolerated |
| N | Neutral | 
| B | Benign | 
| P | Possibly damaging |
